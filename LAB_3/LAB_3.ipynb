```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('lena_gray_256.tif')

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.title('Original Image')
plt.axis('off')
plt.show()
```

```python
height, width = img.shape[:2]

resized_img = cv2.resize(
    img,
    (width * 2, height * 2),
    interpolation=cv2.INTER_CUBIC
)

resized_rgb = cv2.cvtColor(resized_img, cv2.COLOR_BGR2RGB)

plt.imshow(resized_rgb)
plt.title('Increased Image Size')
plt.axis('off')
plt.show()
```

```python
height, width = img.shape[:2]

center = (width // 2, height // 2)

rotation_matrix = cv2.getRotationMatrix2D(
    center,
    120,
    1.0
)

rotated_img = cv2.warpAffine(
    img,
    rotation_matrix,
    (width, height)
)

rotated_rgb = cv2.cvtColor(rotated_img, cv2.COLOR_BGR2RGB)

plt.imshow(rotated_rgb)
plt.title('Rotated Image 120 Degrees')
plt.axis('off')
plt.show()
```

```python
height, width = img.shape[:2]

shear_factor = 0.5

shear_matrix = np.float32([
    [1, shear_factor, 0],
    [0, 1, 0]
])

sheared_img = cv2.warpAffine(
    img,
    shear_matrix,
    (int(width + shear_factor * height), height)
)

sheared_rgb = cv2.cvtColor(sheared_img, cv2.COLOR_BGR2RGB)

plt.imshow(sheared_rgb)
plt.title('Sheared Image')
plt.axis('off')
plt.show()
```

```python
gray_img = cv2.imread(
    'lena_gray_256.tif',
    cv2.IMREAD_GRAYSCALE
)

plt.imshow(gray_img, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')
plt.show()
```

```python
negative_img = 255 - gray_img

plt.imshow(negative_img, cmap='gray')
plt.title('Negative Image')
plt.axis('off')
plt.show()
```

```python
c = 255 / np.log(1 + np.max(gray_img))

log_img = c * np.log(
    1 + gray_img.astype(np.float32)
)

log_img = np.uint8(log_img)

plt.imshow(log_img, cmap='gray')
plt.title('Log Transformation')
plt.axis('off')
plt.show()
```

```python
gamma = 0.5

normalized_img = gray_img / 255.0

power_law_img = np.power(
    normalized_img,
    gamma
)

power_law_img = np.uint8(
    power_law_img * 255
)

plt.imshow(power_law_img, cmap='gray')
plt.title('Power Law Transformation')
plt.axis('off')
plt.show()
```

```python
plt.figure(figsize=(15, 10))

plt.subplot(2, 4, 1)
plt.imshow(img_rgb)
plt.title('Original')
plt.axis('off')

plt.subplot(2, 4, 2)
plt.imshow(resized_rgb)
plt.title('Resize')
plt.axis('off')

plt.subplot(2, 4, 3)
plt.imshow(rotated_rgb)
plt.title('Rotate 120°')
plt.axis('off')

plt.subplot(2, 4, 4)
plt.imshow(sheared_rgb)
plt.title('Shear')
plt.axis('off')

plt.subplot(2, 4, 5)
plt.imshow(gray_img, cmap='gray')
plt.title('Grayscale')
plt.axis('off')

plt.subplot(2, 4, 6)
plt.imshow(negative_img, cmap='gray')
plt.title('Negative')
plt.axis('off')

plt.subplot(2, 4, 7)
plt.imshow(log_img, cmap='gray')
plt.title('Log')
plt.axis('off')

plt.subplot(2, 4, 8)
plt.imshow(power_law_img, cmap='gray')
plt.title('Power Law')
plt.axis('off')

plt.tight_layout()
plt.show()
```
